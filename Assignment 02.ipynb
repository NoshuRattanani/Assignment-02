```python
#Question 10: Job Eligibility System A company wants to filter candidates for a job. 
#    Ask the candidate for: • Age  • Years of experience  • Technical test score  Eligibility rules: 
#    • Age must be at least 18.  
#        • Experience must be at least 1 year.  
#        • Technical test score must be at least 60.  
#            If all requirements are satisfied, display: Candidate is Eligible Otherwise, display the specific reason why the candidate is not eligible 
age = int(input("Enter your age: "))
experience = int(input("Enter your years of experience: "))
score = float(input("Enter your technical test score: "))

if age >= 18 and experience >= 1 and score >= 60:
    print("Candidate is Eligible")
else:
    if age < 18:
        print("Not Eligible: Age must be at least 18")
    if experience < 1:
        print("Not Eligible: Experience must be at least 1 year")
    if score < 60:
        print("Not Eligible: Technical test score must be at least 60")
```
